Lo primero que vamos a hacer es importar las librerias que vamos a utilizar, en este caso, importamos la nativa de google colab para montar el drive
Y por mi parte yo usare numpy para trabajar con arreglos en la practica

```python
from google.colab import drive
import numpy as np
drive.mount('/content/drive')
```

Mandamos a abrir el archivo mbox del drive y lo guardamos en una variable que se llamara mbox

```python
mbox = open('/content/drive/MyDrive/ClassFiles/mbox.txt')
```

Recorremos el arreglo linea por linea buscando las lineas que empiecen con "X-DSPAM-Confidence", como lo pide la practica, lo que yo hice despues fue contar digito por digito donde empezaba la cantidad y guardarlo en una variable llamada num como flotante para manejar los decimales, de ahi guarde num en un arreglo de numpy

Imprimimos las lineas y num para verificar que coincidan

```python
nums = np.array([])
for line in mbox:
  line = line.rstrip()
  if line.startswith('X-DSPAM-Confidence'):
    num = float(line[20:])
    nums = np.append(nums, num)
    print(num)
    print(line)
```

Ya una vez que tenemos nuestro arreglo nums, lo demas es utilizar variables nativas de numpy para sacar los siguientes datos que pide la practica


*   Total de registros
*   Sumatoria de los registros
*   Promedio
*   Varianza



```python
print("Total", nums.size)
print("Suma:", nums.sum())
print("Promedio:", nums.mean())
print("Varianza:", nums.var())
```
