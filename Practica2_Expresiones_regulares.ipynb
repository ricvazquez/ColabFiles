Abrimos el archivo con with y lo guardamos dentro de la variable info, ahora que usamos la funcion with, el archivo se abrio y se cerro son necesidad de hacerlo manualmente

```python
with open('/content/drive/MyDrive/ClassFiles/informe_acciones.txt', 'r', encoding="latin-1") as inf:
  info = inf.readlines()
```

Ahora recorremos el archivo linea por linea, y con la ayuda de expresiones regulares, buscamos donde aparezca la palabra accion de, para saber de que empresa estamos hablando, y despues buscamos la palabra Valor, para encontrar la linea donde tiene el valor de la accion. Una vez hecho esto, lo guardamos en una lista

```python
valores = np.empty((0,2))
valores_list = []
empresa = ""
valor = ""
for line in info:
  line = line.rstrip()
  if re.search('acción de', line):
    empresa = line.split("acción de", 1)[1].strip().rstrip(".")
  if re.search('Valor', line):
    valor = line.split("actual:", 1)[1].strip()
    valores_list.append([empresa, valor])
```

Recorremos la lista que hicimos para imprmir el nombre de la empresa y el valor de la accion y los imprimimos en pantalla

```python
valores = np.array(valores_list)
for line in valores:
  print('La empresa:',line[0],' tiene un valor de accion de:',line[1])
# print(valores)
```
